```python

# ================================================================
# DECODELABS - DATA ANALYTICS PROJECT 2
# RETAIL SALES FORENSICS & CUSTOMER INTELLIGENCE
# Complete End-to-End Exploratory Data Analysis
# ================================================================

# STEP 1: INSTALL REQUIRED LIBRARIES
import subprocess
import sys

subprocess.check_call([
    sys.executable, "-m", "pip", "install", "-q",
    "pandas", "numpy", "matplotlib"
])

# STEP 2: IMPORT LIBRARIES
import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from IPython.display import display, Markdown

warnings.filterwarnings("ignore")
np.random.seed(42)

# STEP 3: CREATE PROJECT FOLDER
OUTPUT = Path("DecodeLabs_Project_2")
OUTPUT.mkdir(exist_ok=True)

print("=" * 65)
print("DECODELABS | PROJECT 2: EXPLORATORY DATA ANALYSIS")
print("=" * 65)

# STEP 4: GENERATE A BUSINESS-RELEVANT DATASET
# This is simulated data, not real company data.

n = 1500

dates = pd.date_range(
    start="2025-01-01",
    end="2025-12-31",
    periods=n
)

categories = [
    "Electronics", "Fashion", "Home & Living",
    "Beauty", "Sports"
]

category = np.random.choice(
    categories, n,
    p=[0.25, 0.25, 0.20, 0.18, 0.12]
)

price_ranges = {
    "Electronics": (100, 1200),
    "Fashion": (15, 200),
    "Home & Living": (30, 500),
    "Beauty": (10, 150),
    "Sports": (20, 350)
}

prices = np.array([
    np.random.uniform(*price_ranges[c])
    for c in category
])

quantity = np.random.randint(1, 8, n)

# Seasonal sales variation
month = dates.month
seasonal_factor = 1 + 0.25 * np.sin(
    2 * np.pi * (month - 1) / 12
)

prices = prices * seasonal_factor

customers = [
    f"CUST_{i:04d}" for i in
    np.random.randint(1, 401, n)
]

regions = np.random.choice(
    ["North", "South", "East", "West"],
    n
)

payment = np.random.choice(
    ["Cash", "Card", "Online Wallet"],
    n,
    p=[0.20, 0.45, 0.35]
)

df = pd.DataFrame({
    "Transaction_ID": [
        f"TRX_{i:05d}" for i in range(1, n + 1)
    ],
    "Date": dates,
    "Customer_ID": customers,
    "Category": category,
    "Region": regions,
    "Quantity": quantity,
    "Unit_Price": np.round(prices, 2),
    "Payment_Method": payment
})

df["Revenue"] = (
    df["Quantity"] * df["Unit_Price"]
).round(2)

# Introduce controlled data-quality problems
missing_idx = np.random.choice(
    df.index, 25, replace=False
)
df.loc[missing_idx[:10], "Category"] = np.nan
df.loc[missing_idx[10:18], "Quantity"] = np.nan
df.loc[missing_idx[18:], "Unit_Price"] = np.nan

# Invalid dates
df.loc[30:33, "Date"] = "INVALID_DATE"

# Suspicious negative values
df.loc[40, "Quantity"] = -5
df.loc[41, "Unit_Price"] = -100

# Introduce extreme but positive transactions
df.loc[50, "Unit_Price"] = 15000
df.loc[51, "Quantity"] = 100

# Duplicate transaction
df = pd.concat(
    [df, df.iloc[[100]]],
    ignore_index=True
)

# Save original evidence
df.to_csv(
    OUTPUT / "raw_retail_dataset.csv",
    index=False
)

print("\nRAW DATASET CREATED")
print("Dataset shape:", df.shape)
display(df.head(10))


# STEP 5: INITIAL DATA INSPECTION

print("\n" + "=" * 65)
print("DATA FORENSICS - INITIAL INVESTIGATION")
print("=" * 65)

print("\nDATA TYPES:")
print(df.dtypes)

print("\nDATASET INFORMATION:")
df.info()

print("\nMISSING VALUES:")
print(df.isnull().sum())

print("\nDUPLICATE RECORDS:")
print(df.duplicated().sum())

print("\nRAW STATISTICAL SUMMARY:")
display(df.describe(include="all").T)


# STEP 6: DATA CLEANING

clean = df.copy()

# Standardize text fields
for col in ["Category", "Region", "Payment_Method"]:
    clean[col] = clean[col].astype("string").str.strip()

# Convert date safely
clean["Date"] = pd.to_datetime(
    clean["Date"], errors="coerce"
)

# Convert numerical columns safely
for col in ["Quantity", "Unit_Price"]:
    clean[col] = pd.to_numeric(
        clean[col], errors="coerce"
    )

# Record missing and invalid values before cleaning
missing_before = clean.isnull().sum()

invalid_dates = clean["Date"].isnull().sum()

duplicate_count = clean.duplicated().sum()

# Remove duplicate rows
clean = clean.drop_duplicates()

# Remove rows with unusable dates or categories
clean = clean.dropna(
    subset=["Date", "Category"]
)

# Remove impossible negative values
clean = clean[
    (clean["Quantity"].isna() |
     (clean["Quantity"] > 0))
    &
    (clean["Unit_Price"].isna() |
     (clean["Unit_Price"] > 0))
].copy()

# Fill missing numerical values using median
clean["Quantity"] = clean["Quantity"].fillna(
    clean["Quantity"].median()
)

clean["Unit_Price"] = clean["Unit_Price"].fillna(
    clean["Unit_Price"].median()
)

# Recalculate revenue rather than trusting raw revenue
clean["Revenue"] = (
    clean["Quantity"] * clean["Unit_Price"]
).round(2)

clean["Month"] = clean["Date"].dt.month
clean["Month_Name"] = clean["Date"].dt.strftime("%B")
clean["Quarter"] = clean["Date"].dt.quarter

clean = clean.reset_index(drop=True)

print("\nCLEANING REPORT")
print("-" * 50)
print("Original records:", len(df))
print("Duplicate records found:", duplicate_count)
print("Invalid dates found:", invalid_dates)
print("Records after cleaning:", len(clean))
print("Records removed:", len(df) - len(clean))

print("\nMISSING VALUES AFTER CLEANING:")
print(clean.isnull().sum())

clean.to_csv(
    OUTPUT / "cleaned_retail_dataset.csv",
    index=False
)


# STEP 7: BASIC DESCRIPTIVE STATISTICS

print("\n" + "=" * 65)
print("DESCRIPTIVE STATISTICS")
print("=" * 65)

numeric_columns = [
    "Quantity", "Unit_Price", "Revenue"
]

statistics = clean[numeric_columns].agg(
    ["count", "mean", "median", "std",
     "min", "max"]
).T

statistics["Q1"] = clean[numeric_columns].quantile(0.25)
statistics["Q3"] = clean[numeric_columns].quantile(0.75)

display(statistics.round(2))

statistics.to_csv(
    OUTPUT / "descriptive_statistics.csv"
)


# STEP 8: FIVE-NUMBER SUMMARY

print("\nFIVE-NUMBER SUMMARY")

five_number = clean[numeric_columns].describe().loc[
    ["min", "25%", "50%", "75%", "max"]
]

display(five_number.round(2))


# STEP 9: DISTRIBUTION AND SKEWNESS ANALYSIS

print("\nDISTRIBUTION ANALYSIS")

for col in numeric_columns:
    skewness = clean[col].skew()

    if skewness > 0.5:
        shape = "Right-skewed distribution"
    elif skewness < -0.5:
        shape = "Left-skewed distribution"
    else:
        shape = "Approximately symmetrical distribution"

    print(
        f"{col}: Skewness = {skewness:.2f} | {shape}"
    )


# STEP 10: OUTLIER DETECTION USING IQR

print("\n" + "=" * 65)
print("OUTLIER INVESTIGATION - IQR METHOD")
print("=" * 65)

outlier_results = {}

for col in numeric_columns:

    Q1 = clean[col].quantile(0.25)
    Q3 = clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = clean[
        (clean[col] < lower_limit) |
        (clean[col] > upper_limit)
    ]

    outlier_results[col] = {
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Limit": lower_limit,
        "Upper_Limit": upper_limit,
        "Outlier_Count": len(outliers)
    }

    print(f"\nColumn: {col}")
    print("Lower limit:", round(lower_limit, 2))
    print("Upper limit:", round(upper_limit, 2))
    print("Outliers found:", len(outliers))

    if len(outliers) > 0:
        print("Sample outliers:")
        display(outliers.head(5))

pd.DataFrame(outlier_results).T.to_csv(
    OUTPUT / "iqr_outlier_report.csv"
)


# STEP 11: Z-SCORE OUTLIER DETECTION

print("\nZ-SCORE OUTLIER ANALYSIS")

zscore_results = {}

for col in numeric_columns:

    std = clean[col].std()

    if std == 0 or pd.isna(std):
        z = pd.Series(0, index=clean.index)
    else:
        z = (
            clean[col] - clean[col].mean()
        ) / std

    flagged = clean[z.abs() > 3]

    zscore_results[col] = len(flagged)

    print(
        f"{col}: {len(flagged)} observations "
        "with absolute Z-score greater than 3"
    )


# STEP 12: MONTHLY SALES TREND

monthly_sales = (
    clean.groupby("Month")["Revenue"]
    .sum()
    .reindex(range(1, 13), fill_value=0)
)

print("\nMONTHLY REVENUE:")
display(monthly_sales.round(2))

monthly_sales.to_csv(
    OUTPUT / "monthly_sales.csv"
)

best_month = monthly_sales.idxmax()
worst_month = monthly_sales.idxmin()

print(
    "Highest revenue month:",
    pd.Timestamp(2025, best_month, 1).strftime("%B")
)

print(
    "Lowest revenue month:",
    pd.Timestamp(2025, worst_month, 1).strftime("%B")
)


# STEP 13: CATEGORY PERFORMANCE

category_sales = (
    clean.groupby("Category")
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Transactions=("Transaction_ID", "count"),
        Average_Revenue=("Revenue", "mean"),
        Total_Quantity=("Quantity", "sum")
    )
    .sort_values("Total_Revenue", ascending=False)
)

category_sales["Revenue_Share_Percent"] = (
    category_sales["Total_Revenue"] /
    clean["Revenue"].sum() * 100
)

print("\nCATEGORY PERFORMANCE:")
display(category_sales.round(2))

category_sales.to_csv(
    OUTPUT / "category_performance.csv"
)


# STEP 14: CUSTOMER SEGMENTATION

customer_analysis = (
    clean.groupby("Customer_ID")
    .agg(
        Total_Spending=("Revenue", "sum"),
        Total_Orders=("Transaction_ID", "count"),
        Average_Order_Value=("Revenue", "mean")
    )
)

spending_threshold = customer_analysis[
    "Total_Spending"
].quantile(0.75)

customer_analysis["Segment"] = np.where(
    customer_analysis["Total_Spending"] >= spending_threshold,
    "High Value",
    np.where(
        customer_analysis["Total_Spending"] >=
        customer_analysis["Total_Spending"].median(),
        "Medium Value",
        "Low Value"
    )
)

segment_summary = (
    customer_analysis.groupby("Segment")
    .agg(
        Customers=("Total_Spending", "count"),
        Revenue=("Total_Spending", "sum"),
        Average_Spending=("Total_Spending", "mean")
    )
)

print("\nCUSTOMER SEGMENTATION:")
display(segment_summary.round(2))

customer_analysis.to_csv(
    OUTPUT / "customer_segmentation.csv"
)


# STEP 15: REGIONAL ANALYSIS

regional_sales = (
    clean.groupby("Region")["Revenue"]
    .agg(["sum", "mean", "count"])
    .sort_values("sum", ascending=False)
)

print("\nREGIONAL SALES:")
display(regional_sales.round(2))


# STEP 16: PAYMENT METHOD ANALYSIS

payment_analysis = (
    clean.groupby("Payment_Method")["Revenue"]
    .agg(["sum", "mean", "count"])
)

print("\nPAYMENT METHOD ANALYSIS:")
display(payment_analysis.round(2))


# STEP 17: CORRELATION ANALYSIS

print("\nCORRELATION MATRIX")

correlation = clean[
    ["Quantity", "Unit_Price", "Revenue"]
].corr(method="pearson")

display(correlation.round(3))

correlation.to_csv(
    OUTPUT / "correlation_matrix.csv"
)


# STEP 18: VISUALIZATIONS

plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.titlesize"] = 14

# Chart 1: Monthly revenue
plt.figure()
plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)
plt.title("Monthly Retail Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(range(1, 13))
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(OUTPUT / "01_monthly_revenue.png", dpi=150)
plt.show()
plt.close()

# Chart 2: Category revenue
plt.figure()
category_sales["Total_Revenue"].sort_values().plot(
    kind="barh"
)
plt.title("Revenue by Product Category")
plt.xlabel("Revenue")
plt.ylabel("Category")
plt.tight_layout()
plt.savefig(OUTPUT / "02_category_revenue.png", dpi=150)
plt.show()
plt.close()

# Chart 3: Revenue distribution
plt.figure()
plt.hist(
    clean["Revenue"],
    bins=35,
    edgecolor="black"
)
plt.title("Distribution of Transaction Revenue")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig(OUTPUT / "03_revenue_distribution.png", dpi=150)
plt.show()
plt.close()

# Chart 4: Boxplot
plt.figure()
plt.boxplot(
    clean["Revenue"],
    vert=False
)
plt.title("Revenue Outlier Investigation - Boxplot")
plt.xlabel("Revenue")
plt.tight_layout()
plt.savefig(OUTPUT / "04_revenue_boxplot.png", dpi=150)
plt.show()
plt.close()

# Chart 5: Correlation heatmap
plt.figure()
plt.imshow(
    correlation,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)
plt.colorbar(label="Pearson Correlation")
plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=25
)
plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

for i in range(len(correlation)):
    for j in range(len(correlation)):
        plt.text(
            j, i,
            f"{correlation.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.title("Correlation Between Numerical Variables")
plt.tight_layout()
plt.savefig(OUTPUT / "05_correlation_heatmap.png", dpi=150)
plt.show()
plt.close()

# Chart 6: Regional sales
plt.figure()
regional_sales["sum"].sort_values().plot(
    kind="bar"
)
plt.title("Regional Revenue Comparison")
plt.xlabel("Region")
plt.ylabel("Revenue")
plt.tight_layout()
plt.savefig(OUTPUT / "06_regional_sales.png", dpi=150)
plt.show()
plt.close()

# Chart 7: Customer segments
plt.figure()
segment_summary["Customers"].plot(
    kind="bar"
)
plt.title("Customer Segmentation")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.savefig(OUTPUT / "07_customer_segments.png", dpi=150)
plt.show()
plt.close()


# STEP 19: BUSINESS INSIGHTS

total_revenue = clean["Revenue"].sum()
total_transactions = len(clean)
average_revenue = clean["Revenue"].mean()
median_revenue = clean["Revenue"].median()
total_customers = clean["Customer_ID"].nunique()

top_category = category_sales.index[0]
top_region = regional_sales.index[0]
top_payment = payment_analysis["sum"].idxmax()

high_value_count = int(
    segment_summary.loc["High Value", "Customers"]
)

high_value_revenue = float(
    segment_summary.loc["High Value", "Revenue"]
)

high_value_share = (
    high_value_revenue / total_revenue * 100
    if total_revenue else 0
)

mean_median_difference = (
    average_revenue - median_revenue
)

print("\n" + "=" * 65)
print("BUSINESS DIAGNOSIS - THE SO WHAT TEST")
print("=" * 65)

insights = [
    f"Total cleaned revenue: {total_revenue:,.2f}",
    f"Valid transactions: {total_transactions:,}",
    f"Unique customers: {total_customers:,}",
    f"Average transaction revenue: {average_revenue:,.2f}",
    f"Median transaction revenue: {median_revenue:,.2f}",
    f"Top revenue category: {top_category}",
    f"Top performing region: {top_region}",
    f"Most revenue-generating payment method: {top_payment}",
    f"Highest revenue month: {pd.Timestamp(2025, best_month, 1).strftime('%B')}",
    f"Lowest revenue month: {pd.Timestamp(2025, worst_month, 1).strftime('%B')}",
    f"High-value customers: {high_value_count}",
    f"High-value customer revenue contribution: {high_value_share:.2f}%"
]

for i, insight in enumerate(insights, 1):
    print(f"{i}. {insight}")


# STEP 20: AUTOMATIC BUSINESS RECOMMENDATIONS

recommendations = [
    f"Prioritize inventory and marketing for {top_category}, "
    "the highest revenue-generating category.",

    f"Study the performance of {top_region} and identify "
    "practices that can be applied to other regions.",

    f"Use targeted loyalty campaigns to retain the "
    f"{high_value_count} high-value customers.",

    "Investigate flagged extreme transactions individually "
    "before treating them as errors.",

    "Plan inventory and promotional campaigns around "
    "the observed monthly sales pattern.",

    "Continue monitoring missing values, invalid dates, "
    "duplicate records and impossible numeric values."
]

print("\nBUSINESS RECOMMENDATIONS:")
for i, recommendation in enumerate(recommendations, 1):
    print(f"{i}. {recommendation}")


# STEP 21: CREATE EXECUTIVE SUMMARY

summary = f"""
# DECODELABS - PROJECT 2
## Retail Sales Forensics & Customer Intelligence

### 1. Problem Statement
The objective is to investigate retail transaction data,
identify data-quality issues, understand sales distributions,
discover trends, analyze customers and generate business insights.

### 2. Methodology
- Dataset generation and initial inspection
- Data-quality and forensic investigation
- Missing-value and duplicate handling
- Invalid date and numerical value treatment
- Descriptive statistical analysis
- Mean, median and five-number summary
- Distribution and skewness analysis
- IQR and Z-score outlier detection
- Monthly and regional trend analysis
- Customer segmentation
- Pearson correlation analysis
- Business interpretation and recommendations

### 3. Dataset
This project uses a simulated dataset containing
{len(df):,} raw records.

Records after cleaning: {len(clean):,}

Duplicate records detected: {duplicate_count}

Invalid dates detected: {invalid_dates}

### 4. Key Findings

Total revenue: {total_revenue:,.2f}

Average transaction revenue: {average_revenue:,.2f}

Median transaction revenue: {median_revenue:,.2f}

Total valid transactions: {total_transactions:,}

Unique customers: {total_customers:,}

Top product category: {top_category}

Top region: {top_region}

Top payment method: {top_payment}

Highest revenue month: {pd.Timestamp(2025, best_month, 1).strftime('%B')}

Lowest revenue month: {pd.Timestamp(2025, worst_month, 1).strftime('%B')}

High-value customer revenue contribution: {high_value_share:.2f}%

### 5. Business Recommendations

""" + "\n".join(
    f"- {r}" for r in recommendations
) + """

### 6. Important Limitations
- The dataset is simulated and does not represent a real retailer.
- Statistical relationships do not establish causation.
- Outliers require contextual investigation before removal.
- Customer segments are based on spending thresholds,
  not a validated predictive model.

### 7. Conclusion
Exploratory Data Analysis transformed raw transaction
records into structured information and business insights.
The investigation demonstrated data cleaning, descriptive
statistics, distribution analysis, anomaly detection,
correlation and stakeholder-oriented reporting.
"""

with open(
    OUTPUT / "EXECUTIVE_SUMMARY.md",
    "w",
    encoding="utf-8"
) as f:
    f.write(summary)


# STEP 22: SAVE FINAL REPORTS

clean.to_excel(
    OUTPUT / "cleaned_retail_dataset.xlsx",
    index=False
)

# Save all important output files
print("\n" + "=" * 65)
print("PROJECT EXECUTION COMPLETED")
print("=" * 65)

print("\nOUTPUT FILES:")

for file in sorted(OUTPUT.iterdir()):
    print("✔", file.name)

print("\nPROJECT FOLDER:", OUTPUT.resolve())

print("\nFINAL DATASET PREVIEW:")
display(clean.head(10))

print("\nEXECUTIVE SUMMARY:")
display(Markdown(summary))

print("\nThank you for completing DecodeLabs Project 2!")
```
